## Initial Findings

The dataset contains 30,000 transaction records across 39 columns and covers the period from January 1, 2024 to December 31, 2026.

Each Order_ID is unique, resulting in 30,000 unique orders. The dataset contains 8,998 unique customers, 10 unique products, and 5 unique stores.

No duplicate rows or missing values were identified during the initial inspection.

The Order_Date column is already stored as a datetime data type, while numeric and categorical fields are appropriately recognized by pandas.

These findings describe the initial structure of the raw dataset. Business-rule validation and data-quality checks will be performed in the next stage.
